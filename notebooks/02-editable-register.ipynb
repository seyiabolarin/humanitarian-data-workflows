## Setup
From the repository root, install `requirements.txt`. Open this notebook there or inside `notebooks/`. No account, token or internet connection is required to run the example after installation.


```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if not (ROOT / "python/workflows.py").is_file():
    raise RuntimeError("Open this notebook from the repository root or notebooks folder.")
sys.path.insert(0, str(ROOT / "python"))
from workflows import (load_register, register_pdf, register_docx, placeholder,
                       compress_folder, prepare_local_attachments, OUTPUT)
rows = load_register()
assert len(rows) == 3  # bundled fictional fixture, not a programme result
```

## Steps

```python
docx = register_docx(rows, OUTPUT / "editable-register.docx")
pdf = register_pdf(rows, OUTPUT / "editable-register.pdf")
assert docx.is_file() and pdf.is_file()
```
